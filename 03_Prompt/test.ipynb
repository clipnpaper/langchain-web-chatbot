### 실습 1. 함수를 Runnable로 바꾸기 (API 키 불필요)

```python
from langchain_core.runnables import RunnableLambda

def add_one(x):
    return x + 1

r_add = RunnableLambda(add_one)
print(r_add.invoke(3))
print(r_add.batch([1,2,3]))
```

### 실습 2. |로 연결하기 (API 키 불필요)

```python
from langchain_core.runnables import RunnableLambda

r_add = RunnableLambda(lambda x: x + 1)
r_double = RunnableLambda(lambda x: x * 2)

chain_a = r_add | r_double
chain_b = r_double | r_add

print(chain_a.invoke(3))
print(chain_b.invoke(3))
```

### 실습 3. RunnableParallel과 RunnablePassthrough (API 키 불필요)

```python
from langchain_core.runnables import (RunnableLambda, RunnableParallel, RunnablePassthrough)

analyzer = RunnableParallel(
    original = RunnablePassthrough(),
    upper = RunnableLambda(lambda x: x.upper()),
    length = RunnableLambda(lambda x: len(x)),
)

print(analyzer.invoke("hello"))
```

**과제 1 (기초, API 키 불필요)** — 문자열을 받아 양쪽 공백을 없애고, 대문자로 바꾼 뒤, 끝에 `!`를 붙이는 체인을 만드세요.

```python
from langchain_core.runnables import RunnableLambda

def wow(x: str):
    return x + "!"

def upper(x: str):
    return x.upper()

def strip(x: str):
    return x.strip()

a = RunnableLambda(strip)
b = RunnableLambda(upper)
c = RunnableLambda(wow)

# a = RunnableLambda(lambda x : x.strip())
# b = RunnableLambda(lambda x : x.upper())
# c = RunnableLambda(lambda x : x + "!")

chain = a | b | c

print(chain.invoke("   hello world  "))
```

```python
from langchain_core.runnables import RunnableLambda

my_list = [10,20,30]

def total(x):
    total = 0
    for num in x:
        total += num
    return total

def average(x):
    total = 0
    for num in x:
        total += num
    return "{:.1f}".format(total / len(x))

def max(x):
    max = 0
    for num in x:
        if max <= num: max = num
    return max

a = RunnableLambda(total)
b = RunnableLambda(average)
c = RunnableLambda(max)

dic = {"total" : a.invoke(my_list), "average": b.invoke(my_list), "max" : c.invoke(my_list)}
print(dic)
```

```python
from langchain_core.runnables import RunnableLambda, RunnableParallel
totalnum = RunnableLambda(lambda x: sum(x))
aver = RunnableLambda(lambda x: average(x))
maximum = RunnableLambda(lambda x: max(x))

chain = RunnableParallel(
    total = totalnum,
    aver = aver,
    max = maximum,
)
print(chain.invoke([10,20,30]))
```
