### 문항 8. 고객 문의 분류기 완성 (6점)
---
▶ 시나리오
- 고객센터에 접수되는 문의를 결제 / 배송 / 환불 / 기타 네 가지로 자동 분류하는 기능을 배포해야한다. 아래 코드의 빈칸(____)을 채워 완성하시오.

````python
# 🔧 LLM 호출 헬퍼 함수 — 앞으로 계속 사용합니다
def ask_llm(messages, temperature=0.3, max_tokens=500):
    """메시지 리스트를 받아서 LLM의 답변 텍스트를 돌려준다."""
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return response.choices[0].message.content
````

````python
def classify_inquiry(text: str) -> str:
    messages = [
        {"role": "system", "content": "고객 문의를 4개 카테고리 중 하나로 분류"
                                        "(결제, 배송, 환불, 기타) 중 하나로 분류하세요"},
        {"role": "user", "content": text},
    ]
    return ask_llm(messages, temperature=0, max_tokens=2)

test_cases = [
    "어제 주문했는데 카드 결제가 두 번 되었어요.",
    "택배가 일주일째 안 와요.",
    "사이즈가 안 맞아서 반품하고 싶어요.",
    "영업시간이 어떻게 되나요?",
]
for text in test_cases:
    category = classify_inquiry(text)
    print(f"[{category:>4}] {text}")
````

---
▶ 시나리오

- temperature 값이 결과에 미치는 영향을 팀원에게 시연해야 한다. 정답이 정해진 질문으로는 차이가 드러나지 않으므로 창의성이 필요한 질문을 사용한다.

▶ 요구사항

해 출력하시오.
2) 출력 아래 마크다운 셀에 관찰한 차이를 두 줄로 쓰시오.


````python
message = [{"role":"user" ,"content": "AI 스타트업 이름 3개를 지어주세요"}]

print(ask_llm(messages=message, temperature=0))
print(ask_llm(messages=message, temperature=0))
print(ask_llm(messages=message, temperature=1.2))
print(ask_llm(messages=message, temperature=1.2))
````

temperature=0의 두답변은 차이가 거의 없는 답변을 내세웠고,
temperature=1.2의 두답변은 창의성이 돋보여 두 답변이 거의 다른 이름을 추천해줬다.

````python
PRICE_INPUT = 0.15   # 입력
PRICE_OUTPUT = 0.60  # 출력 (출력이 4배 비쌈!)
USD_TO_KRW = 1400    # 환율 (대략)

def cost_korean_won(usage):
    """usage 객체를 받아 비용을 원화로 환산"""
    input_cost = usage.prompt_tokens / 1_000_000 * PRICE_INPUT
    output_cost = usage.completion_tokens / 1_000_000 * PRICE_OUTPUT
    return (input_cost + output_cost) * USD_TO_KRW
````

````python
# 짧은 호출과 긴 호출의 비용 비교
short_resp = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "Hi"}],
)

long_resp = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "Python의 장점을 5가지 설명해주세요."}],
)

for label, resp in [("짧은 호출", short_resp), ("긴 호출", long_resp)]:
    u = resp.usage
    print(f"[{label}] 입력 {u.prompt_tokens} + 출력 {u.completion_tokens} = "
          f"총 {u.total_tokens} 토큰, 비용 약 {cost_korean_won(u):.4f}원")
````
